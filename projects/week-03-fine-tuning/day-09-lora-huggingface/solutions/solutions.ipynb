### Answer key
1. `r` ↔ `r`, `lora_alpha` ↔ `alpha`. Effective scale `= lora_alpha / r = 16/8 = 2.0`.
2. So the loss is computed only over the completion; otherwise ~half the gradient trains the
   model to generate ticket text, wasting capacity and blurring the target behaviour.
3. Duplicates (hash `(prompt,completion)`); leakage (`test.prompt in train.prompts`); format
   drift (regex-match completions to the template); label noise (hand-check a random 50);
   imbalance (class counts + length histogram). Any three.
4. Into the frozen base model — those parameters have `requires_grad=False`; only the LoRA
   `A`/`B` matrices on the target modules are trained.
5. It folds `B·A` into the base weights and returns a plain `transformers` model with no PEFT
   wrapper — so inference has zero extra matmul and you can serve/export it like any model.
6. Train/test leakage — the eval set overlaps training, so 0.98 measures memorisation. The
   0.80 is the real generalisation number.
7. It loads the base model's weights in 4-bit precision (≈1/4 the memory) while the LoRA
   adapter stays in higher precision — this is QLoRA, and it's what lets multi-billion-param
   models fine-tune on a single consumer GPU.
